# Combine Dataset

```python
keeper_cols = [(  'Performance',  'SoTA'),
            (  'Performance',    'GA'),
            (  'Performance', 'Saves'),
            (  'Performance', 'Save%'),
            (  'Performance',    'CS'),
            ('Penalty Kicks', 'PKatt'),
            ('Penalty Kicks',   'PKA'),
            ('Penalty Kicks',  'PKsv'),
            ('Penalty Kicks',   'PKm')]

misc_cols = [( 'Performance',  'CrdY'),
            ( 'Performance',  'CrdR'),
            ( 'Performance', '2CrdY'),
            ( 'Performance',   'Fls'),
            ( 'Performance',   'Fld'),
            ( 'Performance',   'Off'),
            ( 'Performance',   'Crs'),
            ( 'Performance',   'Int'),
            ( 'Performance',  'TklW'),
            ( 'Performance', 'PKwon'),
            ( 'Performance', 'PKcon')]

schedule_cols = ["Poss", "Captain", "Formation", "Opp Formation"]
```

```python
# join keeper
df = shooting.join(keeper[keeper_cols])

# join misc
df = df.join(misc[misc_cols])

# hack way to convert 2-level columns to 1-level columns
df.columns = ['_'.join(filter(None, col)).strip() for col in df.columns]
```

```python
# join schedule
df = df.join(schedule[schedule_cols])
```

```python
df.head(2)
```

# Data Analysis

### Create Target Variable / Y columns

```python
df["win"] = (df["result"] == "W").astype(int)
df["goal_diff"] = df["GF"] - df["GA"]
```

### Feature Exploration

```python
features = [
    "Standard_Sh",       # shots
    "Standard_SoT",      # shots on target
    "Standard_SoT%",     # shot accuracy
    "Poss",              # possession
    "Performance_SoTA",  # opponent shots on target
    "Performance_CrdY",  # yellow cards
    "Performance_CrdR",  # red cards
    "Performance_Int",   # interceptions
    "Performance_TklW",  # tackles won
]
```

```python
df.groupby("win")[features].mean().T
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(
    data=df,
    x="win",
    y="Standard_SoT"
)

plt.xlabel("Match Result")
plt.ylabel("Shots on Target")
plt.title("Shots on Target vs. Match Result")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Standard_SoT",
    y="goal_diff"
)

plt.xlabel("Shots on Target")
plt.ylabel("Goal Difference")
plt.title("Shots on Target vs. Goal Difference")
plt.axhline(0, linestyle="--")
plt.show()
```

### Correlation

```python
features = [
    "Standard_Sh",
    "Standard_SoT",
    "Standard_SoT%",
    "Poss",
    "Performance_SoTA",
    "Performance_CrdY",
    "Performance_CrdR",
    "Performance_Fls",
    "Performance_Off",
    "Performance_Crs",
    "Performance_Int",
    "Performance_TklW"
]

corr = (
    df[features + ["goal_diff"]]
    .corr()["goal_diff"]
    .sort_values()
)

print(corr)
```

```python
plt.figure(figsize=(8, 6))

corr.drop("goal_diff").plot(kind="barh")

plt.xlabel("Correlation with Goal Difference")
plt.ylabel("Feature")
plt.title("Which Statistics Are Related to Goal Difference?")
plt.axvline(0, linestyle="--")
plt.show()
```
