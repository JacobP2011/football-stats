# Load packages

```python
import soccerdata as sd
import pandas as pd

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

# Load Dataset

```python
fb = sd.FBref('ESP-La Liga', '2025')
```

```python
shooting = fb.read_team_match_stats(stat_type="shooting", team="Real Madrid", force_cache=False)
```

```python
keeper = fb.read_team_match_stats(stat_type="keeper", team="Real Madrid")
```

```python
schedule = fb.read_team_match_stats(stat_type="schedule", team="Real Madrid")
```

```python
misc = fb.read_team_match_stats(stat_type="misc", team="Real Madrid")
```

```python
keeper_cols = [(  'Performance',  'SoTA'),
            (  'Performance',    'GA'),
            (  'Performance', 'Saves'),
            (  'Performance', 'Save%'),
            (  'Performance',    'CS'),
            ('Penalty Kicks', 'PKatt'),
            ('Penalty Kicks',   'PKA'),
            ('Penalty Kicks',  'PKsv'),
            ('Penalty Kicks',   'PKm')]

misc_cols = [( 'Performance',  'CrdY'),
            ( 'Performance',  'CrdR'),
            ( 'Performance', '2CrdY'),
            ( 'Performance',   'Fls'),
            ( 'Performance',   'Fld'),
            ( 'Performance',   'Off'),
            ( 'Performance',   'Crs'),
            ( 'Performance',   'Int'),
            ( 'Performance',  'TklW'),
            ( 'Performance', 'PKwon'),
            ( 'Performance', 'PKcon')]

schedule_cols = ["Poss", "Captain", "Formation", "Opp Formation"]
```

```python
# join keeper
df = shooting.join(keeper[keeper_cols])

# join misc
df = df.join(misc[misc_cols])

# hack way to convert 2-level columns to 1-level columns
df.columns = ['_'.join(filter(None, col)).strip() for col in df.columns]
```

```python
# join schedule
df = df.join(schedule[schedule_cols])
```
