# Load Packages

```python
import soccerdata as sd
import pandas as pd

pd.set_option('display.max_columns', None)
```

```python
fb = sd.FBref('ESP-La Liga', '2025')
```

# Load Dataset

```python
keeper = fb.read_team_match_stats(stat_type="keeper", team="Real Madrid", force_cache=False)
```

```python
misc = fb.read_team_match_stats(stat_type="misc", team="Real Madrid", force_cache=False)
```

```python
schedule = fb.read_team_match_stats(stat_type="schedule", team="Real Madrid", force_cache=False)
```

```python
shooting = fb.read_team_match_stats(stat_type="shooting", team="Real Madrid", force_cache=False)
```

```python
shooting2 = shooting.copy()
```

```python
shooting2.tail()
```

# Create Columns

```python
shooting2["is_w"] = shooting2["result"] == "W"
```

```python
shooting2["after_new_coach"] = shooting2["date"] > "2026-01-12"
```

```python
shooting2["goal_diff"] = shooting2["GF"] - shooting2["GA"]
```

```python
shooting2["goal_diff"].value_counts().sum()
```

```python
shooting2.groupby("after_new_coach")["is_w"].mean()
```

```python
shooting.groupby("venue")["GF"].mean()
```

```python
shooting.groupby("day")["GA"].count()`
```
